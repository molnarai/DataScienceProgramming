# Introduction to the Python Programming Language

## 1. What Is Python?

Python is a high-level, general-purpose programming language created by Guido van Rossum and first released in 1991. It was designed around a simple premise: code is read far more often than it is written, so the language should optimize for clarity.

Key characteristics:

- **Interpreted** — code runs directly through an interpreter; no separate compilation step is required.
- **Dynamically typed** — variables do not need type declarations; types are checked at run time.
- **Automatic memory management** — a garbage collector reclaims unused objects.
- **Multi-paradigm** — supports procedural, object-oriented, and functional styles.
- **Batteries included** — a large standard library ships with the language.
- **Cross-platform** — the same source runs on Linux, macOS, and Windows.

The current major version is Python 3. Python 2 reached end of life in 2020 and should not be used for new work.

## 7. Built-in Collections

| Type | Literal | Ordered | Mutable | Typical use |
|---|---|---|---|---|
| `list` | `[1, 2, 3]` | yes | yes | sequence of items that changes |
| `tuple` | `(1, 2, 3)` | yes | no | fixed record, dictionary key |
| `dict` | `{"a": 1}` | yes (insertion) | yes | key → value lookup |
| `set` | `{1, 2, 3}` | no | yes | membership, deduplication |

````python
scores = [88, 92, 79]
scores.append(95)
scores[0] = 90
print(len(scores), max(scores), sum(scores) / len(scores))

point = (3.5, 7.2)          # immutable pair
x, y = point                # tuple unpacking

student = {"name": "Ada", "year": 2, "gpa": 3.9}
student["gpa"] = 4.0
print(student.get("major", "undeclared"))   # default if key is missing

for key, value in student.items():
    print(key, "->", value)

unique_years = {2, 3, 2, 4}   # {2, 3, 4}
````

## 8. Control Flow

### Conditionals

````python
score = 87

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "F"
````

### Loops

`for` iterates over any iterable; `while` repeats until a condition fails.

````python
for city in ["Atlanta", "Boston", "Chicago"]:
    print(city)

for i in range(5):          # 0, 1, 2, 3, 4
    print(i)

for index, city in enumerate(["Atlanta", "Boston"]):
    print(index, city)

count = 3
while count > 0:
    print(count)
    count -= 1
````

`break` exits a loop early; `continue` skips to the next iteration.

### Comprehensions

A compact way to build a collection from an iterable:

````python
squares = [n ** 2 for n in range(10)]
evens = [n for n in range(20) if n % 2 == 0]
lengths = {word: len(word) for word in ["alpha", "beta"]}
````

## 9. Functions

````python
def celsius_to_fahrenheit(celsius):
    """Convert a Celsius temperature to Fahrenheit."""
    return celsius * 9 / 5 + 32


def greet(name, greeting="Hello"):      # default argument
    return f"{greeting}, {name}!"


print(celsius_to_fahrenheit(100))       # 212.0
print(greet("Ada"))                     # Hello, Ada!
print(greet("Ada", greeting="Hi"))      # keyword argument
````

Functions may return multiple values as a tuple:

````python
def min_max(values):
    return min(values), max(values)

low, high = min_max([4, 9, 1, 7])
````

Optional type hints document intent and are checked by external tools, not at run time:

````python
def average(values: list[float]) -> float:
    return sum(values) / len(values)
````

## 11. Files and Errors

Read and write files with a `with` block, which closes the file automatically:

````python
from pathlib import Path

# Create the file this example reads (see the setup cell at the top).
Path("data.txt").write_text("first line\nsecond line\n", encoding="utf-8")

with open("data.txt", "r") as f:
    for line in f:
        print(line.strip())

with open("output.txt", "w") as f:
    f.write("Result: 42\n")

print(Path("output.txt").read_text())
````

Handle failures explicitly rather than letting the program crash:

````python
# `input()` would block a notebook waiting for the keyboard, so this cell uses a
# fixed string in its place. In a script, the first line reads:
#     value = int(input("Enter a number: "))
entered = "42"

try:
    value = int(entered)
except ValueError:
    print("That was not a whole number.")
else:
    print(f"You entered {value}.")
finally:
    print("Done.")
````

Try changing `entered` to something like `"twelve"` and re-running the cell to see the `except` branch and the `finally` branch both fire.

## 12. Objects and Classes

Everything in Python is an object. You can define your own types:

````python
class Student:
    def __init__(self, name, gpa):
        self.name = name
        self.gpa = gpa

    def honors(self):
        return self.gpa >= 3.5

    def __repr__(self):
        return f"Student({self.name!r}, {self.gpa})"


ada = Student("Ada", 3.9)
print(ada.honors())     # True
print(ada)              # Student('Ada', 3.9)
````

## 13. Style Conventions

Python's style guide is **PEP 8**. The main points:

- `snake_case` for variables and functions, `CapWords` for classes, `UPPER_CASE` for constants.
- Four spaces per indentation level.
- Lines under roughly 79–100 characters.
- Two blank lines between top-level definitions.
- Descriptive names: `average_score`, not `avg_s` or `x`.

Tools such as `black` (formatter) and `ruff` or `flake8` (linters) enforce these automatically.

## 14. A Short Complete Example

````python
"""Summarize exam scores read from a text file."""

from pathlib import Path
from statistics import mean, median

# Create the input file this script reads (see the setup cell at the top).
Path("scores.txt").write_text("88\n92\nnot-a-score\n79\n95\n", encoding="utf-8")


def load_scores(path):
    """Return a list of numeric scores, skipping unreadable lines."""
    scores = []
    with open(path) as f:
        for line in f:
            line = line.strip()
            if not line:
                continue
            try:
                scores.append(float(line))
            except ValueError:
                print(f"Skipping invalid line: {line!r}")
    return scores


def summarize(scores):
    if not scores:
        return "No scores available."
    return (
        f"n = {len(scores)}\n"
        f"mean = {mean(scores):.2f}\n"
        f"median = {median(scores):.2f}\n"
        f"range = {min(scores)}-{max(scores)}"
    )


def main():
    scores = load_scores("scores.txt")
    print(summarize(scores))


main()
````
