## 2. Creating a notebook

In JupyterLab: **File → New → Notebook**, then pick the **Python 3
(ipykernel)** kernel when prompted.

A few pieces of vocabulary you will hear all term:

- **Notebook** — the `.ipynb` file. It stores your cells *and* the output they
  produced the last time they ran.
- **Kernel** — the Python process running behind the notebook. It holds every
  variable you have defined. The notebook file is what you see; the kernel is
  what remembers.
- **Cell** — one editable block. Cells run one at a time, and each one runs
  inside that same kernel.

That last point is the whole reason the rest of this notebook exists: the file
on disk and the kernel's memory are **two different things**, and they can
disagree.

## 4. Execution order and the execution counter

Look to the left of a code cell. Before it runs it shows `[ ]`. While it is
running it shows `[*]`. After it finishes it shows a **number**: `[1]`, `[2]`,
and so on.

That number is the **execution counter**. It does not say which cell this is —
it says *how many cells have been executed in this kernel so far*. It is the
only visible trace of something otherwise invisible: the kernel's **hidden
state**.

Run the next three cells in order and watch the numbers.

````python
x = 10
x
````

````python
y = x * 3
y
````

````python
x + y
````

Now the important part. Run the cell below **once** and note the answer.
Then run the *same cell* again. And again.

````python
x = x + 5
x
````

The cell never changed, but the answer did — 15, then 20, then 25. The cell is
not a fact about `x`; it is an *instruction* that modifies whatever `x` happens
to be right now.

Notice too that the execution counter jumped each time, while the counters on
the cells above it stayed put. Those cells are now showing output computed from
an **older value of `x`** than the one currently in the kernel. The notebook on
screen is no longer a truthful record of the kernel's state.

This is the single most common source of confusion for people new to notebooks:

> A notebook can display results that no longer follow from the code above them.

Confirm the disagreement for yourself. Scroll up to the cell that printed `10`,
then run the cell below and compare.

````python
x
````

### Restart and Run All

The cure is to rebuild the kernel's state from nothing:

**Kernel → Restart Kernel and Run All Cells…**

This clears every variable, then executes the cells top to bottom. Afterwards
the counters read `[1]`, `[2]`, `[3]`, … straight down the notebook, and what
you see is exactly what the code produces.

**Do this before you submit any assignment.** A notebook that only works because
of some cell you ran an hour ago and then deleted will fail automated grading.

You can see the kernel's current state at any time with `%whos`, a *magic
command* — a notebook-level instruction, not Python.

## 5. `print` versus automatic display

Jupyter automatically displays the value of the **last expression** in a cell.
It is a convenience, and it has three edges worth knowing.

**Only the last one.** The cell below computes three values and shows you one.

````python
2 + 2
3 + 3
4 + 4
````

**`print` shows all of them**, because printing is an action that happens as the
line runs, not a value returned at the end.

````python
print(2 + 2)
print(3 + 3)
print(4 + 4)
````

**Assignments display nothing.** An assignment is a statement, not an
expression — it has no value to show. The cell below produces no output at all,
even though it definitely did something.

````python
greeting = "Hello, IFI 8410"
````

Name the variable on its own line and you get the automatic display back:

````python
greeting
````

The two are not the same output. Automatic display shows the `repr` — the
developer-facing form, with quotes for strings. `print` shows the `str` — the
human-facing form.

````python
print(greeting)
````

The difference is easy to miss with a plain string and impossible to miss once
there is a newline in it. Automatic display shows the escape sequence `\n`
literally; `print` renders it as an actual line break.

````python
two_lines = "first line\nsecond line"
two_lines
````

````python
print(two_lines)
````

One consequence catches everyone once: `print` **returns nothing**. Its return
value is `None`, so the last line of a cell should be a bare expression, not a
`print`, when you want the rich display Jupyter gives to tables and figures
later in the course.

````python
result = print("printed")
result is None
````

**Rule of thumb:** use the automatic display when you want to inspect one final
value; use `print` when you want to see several things, or to label output.

### 6.1 Arithmetic

Python's arithmetic operators mostly look like ordinary maths. Two do not:
`//` is floor division (divide and discard the remainder) and `%` is modulo
(the remainder itself).

````python
7 + 3
7 - 3
7 * 3
7 / 3
````

Only `7 / 3` was displayed — the automatic display shows the last expression
only. Use `print` to see them together, and note that `/` always produces a
float, even when the division comes out even.

````python
print(7 + 3)
print(7 - 3)
print(7 * 3)
print(7 / 3)
print(7 // 3)
print(7 % 3)
print(7 ** 3)
print(6 / 3)
````

Precedence follows the usual rules — `**` before `*` and `/`, which come before
`+` and `-` — and parentheses override them. When in doubt, add parentheses:
they cost nothing and they document your intent.

````python
print(2 + 3 * 4)
print((2 + 3) * 4)
````

### 6.2 Strings

A string is text in single or double quotes — the two are interchangeable, so
pick whichever avoids escaping the quotes inside your text.

````python
course = "IFI 8410"
topic = 'Introduction to Python'
print(course)
print(topic)
````

`+` joins strings; `*` repeats them. `len` reports the number of characters.
Note that `+` between a string and a number is an error, not an
implicit conversion — Python refuses to guess what you meant.

````python
print(course + ": " + topic)
print("-" * 40)
print(len(course))
````

Methods are functions that belong to a value; call them with a dot. Strings are
**immutable**, so these return a *new* string and leave the original alone.

````python
print(topic.upper())
print(topic.lower())
print(topic.replace("Python", "Data Science"))
print(topic)
````

**f-strings** are the readable way to build text out of values. Put an `f`
before the quote and write expressions inside `{}`.

````python
session = 1
f"Session {session} of {course}: {topic}"
````

Any expression works inside the braces, and a format specifier after a colon
controls how numbers are rendered — `:.2f` means two decimal places.

````python
price = 86.4
people = 3
f"Each of the {people} of us owes ${price / people:.2f}"
````

### 6.3 Variables and assignment

`=` is **assignment**, not equality. It binds a name on the left to the value of
the expression on the right. Read `total = 40 + 2` as *"let `total` refer to
42"*.

A useful mental model: a variable is a **name tag attached to a value**, not a
box containing one. Assigning again moves the tag; it does not change the old
value.

````python
units = 40
price_per_unit = 2.5
total = units * price_per_unit
total
````

Names can be rebound at any time, including to a value of an entirely different
type. Python is **dynamically typed** — the value carries the type, the name
does not.

````python
value = 42
print(type(value))

value = "forty-two"
print(type(value))
````

Because assignment evaluates the right side *first*, this idiom is safe and
very common — it is the same pattern you saw cause surprises in Section 4.

````python
counter = 0
counter = counter + 1
counter += 1
counter
````

Naming is part of correctness, not decoration. The convention in Python is
`lower_case_with_underscores`, and the name should say what the value *means*:

| Avoid | Prefer |
|---|---|
| `x`, `d`, `tmp` | `total_sales`, `days_elapsed` |
| `list`, `sum`, `str` | `student_list`, `row_sum` |
| `thing1`, `thing2` | `first_quarter`, `second_quarter` |

The middle row matters more than it looks: `list`, `sum`, `str`, `type`, and
`print` are all built-in names. Assigning to one shadows the original for the
rest of the session, and the error it eventually causes will point somewhere
else entirely.

## 7. Saving your work and the course naming convention

Save with `Ctrl+S` (`Cmd+S` on macOS), or **File → Save Notebook**. Jupyter also
autosaves periodically, but do not rely on it.

Before you save anything you intend to submit, run
**Kernel → Restart Kernel and Run All Cells…** and check that it completes
without errors. That is what makes the saved output trustworthy.

Notebook filenames in this course follow one pattern:

```
Session_<n>_<ShortTitle>.ipynb        e.g.  Session_1_Introduction.ipynb
Homework_<n>.ipynb                    e.g.  Homework_1.ipynb
```

and the rules behind it:

- **No spaces.** Use underscores. Spaces break shell commands and URLs.
- **Letters, digits, and underscores only** — no `?`, `*`, `#`, `&`, or accents.
- **Case matters.** The grading server is Linux, where `Homework_1.ipynb` and
  `homework_1.ipynb` are different files. macOS and Windows will let you get
  this wrong locally and only fail once submitted.
- **Keep the `.ipynb` extension.**
- **Save it in the assignment's own directory**, not your home directory or your
  Downloads folder.

## Summary

- A notebook is **cells plus saved output**; the **kernel** is what actually
  holds your variables. They can disagree.
- The **execution counter** records how many cells the kernel has run, not
  where you are in the file. Out-of-order execution is the usual cause of
  results that "make no sense".
- **Restart Kernel and Run All Cells** before every submission.
- A cell automatically displays its **last expression** only, and shows nothing
  for an assignment. `print` shows anything you ask it to, whenever it runs.
- Automatic display shows `repr`; `print` shows `str`. The difference is visible
  the moment a string contains `\n`.
- Variables are **names bound to values**. Clear names are part of correct code.
