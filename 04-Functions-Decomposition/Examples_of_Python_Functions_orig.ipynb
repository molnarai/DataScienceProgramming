## Example 1 &mdash; A function with one parameter

**Difficulty:** easy &nbsp;·&nbsp; **Teaches:** `def`, a parameter, `return`

The smallest useful function: take one input, give back one result.

Read the three moving parts:

- `name` is the **parameter** &mdash; a placeholder filled in at each call.
- The f-string builds a new string from it.
- `return` hands that string **back to the caller**, so it can be stored in a variable.

The function does *not* print. That is deliberate: a function that returns can be used in
more places than one that prints. Example 2 makes the difference concrete.

````python
def greet(name: str) -> str:
    """Return a greeting for name."""
    return f"Hello, {name}!"


# The result comes back as a value, so we can store it...
message = greet("Amina")
print(message)

# ...use it directly...
print(greet("Bo"))

# ...or feed it straight into another function.
print(greet("Chen").upper())
print(len(greet("Dee")))
````

## Example 2 &mdash; Return versus print

**Difficulty:** easy &nbsp;·&nbsp; **Teaches:** why `return` is not `print`

These two functions look almost identical, and behave completely differently.

`print()` sends text to the screen for a **human**. `return` sends a value back to the
**program**. A function with no `return` gives back `None`, which is why the second
`print` below shows `None`, and why arithmetic on the result fails.

The rule: **return the answer; let the caller decide whether to print it.**

````python
def show_double(n):          # prints, returns nothing
    print(n * 2)

def double(n):               # returns, prints nothing
    return n * 2


print("show_double(5) displays:")
captured = show_double(5)
print("...and hands back:", captured)      # None

print()
print("double(5) hands back:", double(5))
print("so it composes:", double(double(5)) + 1)

try:
    show_double(5) + 1
except TypeError as error:
    print("but show_double does not:", f"TypeError: {error}")
````

## Example 3 &mdash; A predicate: returning `True` or `False`

**Difficulty:** easy &nbsp;·&nbsp; **Teaches:** returning a Boolean, naming conventions

A function that answers a yes/no question is called a **predicate**. Two conventions:

- Name it as a question: `is_even`, `has_discount`, `looks_valid`.
- Return the comparison **directly**. `return n % 2 == 0` is already a Boolean, so writing
  `if n % 2 == 0: return True else: return False` just adds noise.

Because the result is a real Boolean, it slots straight into `if`, `filter`, and `all`.

````python
def is_even(n: int) -> bool:
    """Return True when n is divisible by 2."""
    return n % 2 == 0


print(is_even(4), is_even(7))

# A predicate is useful everywhere a condition is expected:
print([n for n in range(10) if is_even(n)])
print(all(is_even(n) for n in [2, 4, 6]))

if is_even(10):
    print("10 is even")
````

## Example 4 &mdash; A default parameter value

**Difficulty:** easy &nbsp;·&nbsp; **Teaches:** optional parameters, named arguments

Giving a parameter a default makes it **optional**. `rectangle_area(5)` works because
`height` falls back to `1`.

Two things to notice:

- Parameters **with** defaults must come **after** those without.
- Passing the value **by name** (`height=3`) is clearer than by position at the call site,
  especially once a function has several optional parameters.

````python
def rectangle_area(width: float, height: float = 1.0) -> float:
    """Return the area of a rectangle. Height defaults to 1.0 (a unit strip)."""
    return width * height


print(rectangle_area(5))            # uses the default
print(rectangle_area(5, 3))         # positional override
print(rectangle_area(5, height=3))  # named override -- clearest
print(rectangle_area(width=5, height=3))
````

## Example 5 &mdash; Validating input before computing

**Difficulty:** easy &nbsp;·&nbsp; **Teaches:** `raise`, `ValueError`, failing early

A function should decide what **invalid input** means, and say so out loud.

Here the checks run **before** the arithmetic, so a bad call fails immediately with a clear
message instead of returning a plausible-looking wrong number. Without the check,
`apply_discount(10, 1.5)` would quietly return `-5.00` &mdash; negative revenue that could
travel a long way through a report before anyone noticed.

The `Raises:` line in the docstring is part of the contract: it tells callers what to
expect.

````python
def apply_discount(price: float, discount: float = 0.0) -> float:
    """Return price after a fractional discount.

    Args:
        price: A non-negative price.
        discount: A fraction from 0.0 through 1.0.

    Returns:
        The discounted price.

    Raises:
        ValueError: If price is negative or discount is outside [0, 1].
    """
    if price < 0:
        raise ValueError("price cannot be negative")
    if not 0.0 <= discount <= 1.0:
        raise ValueError("discount must be between 0.0 and 1.0")
    return round(price * (1 - discount), 2)


print(apply_discount(3.50, 0.10))
print(apply_discount(3.50))          # no discount

for bad in [(3.50, 1.5), (-1.0, 0.1)]:
    try:
        apply_discount(*bad)
    except ValueError as error:
        print(f"apply_discount{bad} -> ValueError: {error}")
````

## Example 6 &mdash; Several steps inside one function

**Difficulty:** easy &nbsp;·&nbsp; **Teaches:** local variables, building a result step by step

A function body can hold as many steps as it needs. The names created inside &mdash;
`cleaned`, `parts`, `letters` &mdash; are **local**: they exist only during the call and
cannot collide with anything outside.

That isolation is what makes a function safe to reuse. Each call starts with a fresh set
of local names.

````python
def initials(full_name: str) -> str:
    """Return the uppercase initials of a person's name.

    Args:
        full_name: A name such as "ada lovelace".

    Returns:
        The initials, e.g. "A.L."
    """
    cleaned = full_name.strip()
    parts = cleaned.split()
    letters = [part[0].upper() for part in parts]
    return ".".join(letters) + "."


print(initials("ada lovelace"))
print(initials("  grace brewster murray hopper  "))
print(initials("Katherine Johnson"))

# The locals are gone once the call ends:
try:
    print(parts)
except NameError as error:
    print(f"NameError: {error}")
````

## Example 7 &mdash; Looping over a list inside a function

**Difficulty:** easy &nbsp;·&nbsp; **Teaches:** the accumulator pattern

The **accumulator pattern**: start with an empty total, add to it in a loop, return it at
the end.

The key detail is *where* `total = 0.0` sits. Because it is **inside** the function, every
call starts from zero. If it lived outside as a global, results would pile up across calls
&mdash; the classic notebook bug where re-running a cell changes the answer.

Python's built-in `sum()` does this for you; the explicit loop is shown because the pattern
generalizes to cases `sum()` cannot handle.

````python
def total_price(prices: list[float]) -> float:
    """Return the sum of a list of prices."""
    total = 0.0              # INSIDE the function -- fresh on every call
    for price in prices:
        total += price
    return round(total, 2)


drinks = [3.50, 2.75, 3.50]
print(total_price(drinks))
print(total_price(drinks))       # same answer every time
print(total_price([]))           # empty list -> 0.0

# The built-in does the same thing for this simple case:
print(round(sum(drinks), 2))
````

## Example 8 &mdash; Building a dictionary as the result

**Difficulty:** easy&ndash;medium &nbsp;·&nbsp; **Teaches:** returning a `dict`, `.get()` with a default

Counting is so common it has a standard shape:

```python
counts[key] = counts.get(key, 0) + 1
```

`counts.get(key, 0)` returns the current count, or `0` when the key has not been seen yet.
That avoids a `KeyError` on the first occurrence without needing an `if`.

Returning a **dictionary** rather than printing means the caller can sort it, look up one
value, or feed it onward.

````python
def count_words(text: str) -> dict[str, int]:
    """Return how many times each word appears in text, case-insensitively."""
    counts = {}
    for word in text.lower().split():
        counts[word] = counts.get(word, 0) + 1
    return counts


result = count_words("the cart sells the coffee and the tea")
print(result)

# Because it RETURNS a dict, the caller can do more with it:
print("most common:", max(result, key=result.get))
print("'the' appears", result["the"], "times")
for word, n in sorted(result.items(), key=lambda pair: -pair[1])[:3]:
    print(f"  {word:<8}{n}")
````

## Example 9 &mdash; Handling the empty case

**Difficulty:** easy&ndash;medium &nbsp;·&nbsp; **Teaches:** edge cases, returning `None` deliberately

`sum(values) / len(values)` crashes on an empty list. Every function that divides by a
count has to answer: *what does "no data" mean?*

Two respectable answers:

- **Return `None`** and document it &mdash; good when empty input is normal (a filter matched
  nothing).
- **Raise `ValueError`** &mdash; good when empty input means something upstream is broken.

What is *not* acceptable is letting `ZeroDivisionError` escape with no explanation. Pick a
policy, write it in the docstring, and be consistent.

````python
def mean(values: list[float]) -> float | None:
    """Return the arithmetic mean, or None when values is empty.

    Returning None -- rather than raising -- because an empty selection is a
    normal outcome when filtering a table.
    """
    if not values:
        return None
    return sum(values) / len(values)


print(mean([2.0, 4.0, 6.0]))
print(mean([]))              # None, not a crash

# The annotation `float | None` warns the caller to check before doing arithmetic:
result = mean([])
if result is None:
    print("no data to average")
else:
    print(round(result, 2))

# Compare: the unguarded version.
try:
    sum([]) / len([])
except ZeroDivisionError as error:
    print(f"unguarded -> ZeroDivisionError: {error}")
````

## Example 10 &mdash; Returning more than one value

**Difficulty:** easy&ndash;medium &nbsp;·&nbsp; **Teaches:** tuple return and unpacking

Python has no "multiple return values". `return low, high` builds **one tuple**, and the
call site **unpacks** it into two names.

Knowing it is really a tuple explains everything else: you can store it whole, unpack it,
index it, or ignore part of it with the conventional `_`.

````python
def min_max(values: list[float]) -> tuple[float, float]:
    """Return the (lowest, highest) values."""
    return min(values), max(values)        # parentheses optional -- it is a tuple


prices = [row["price"] for row in sales]

whole = min_max(prices)
print("as a tuple:", whole, type(whole).__name__)

low, high = min_max(prices)                # unpacking
print("unpacked  :", low, high)

print("spread    :", round(high - low, 2))

_, highest = min_max(prices)               # _ means "not needed"
print("just high :", highest)
````

## Example 11 &mdash; Filtering a table

**Difficulty:** medium &nbsp;·&nbsp; **Teaches:** working with a list of dictionaries, returning a new list

The table is a **list of dictionaries**: each row is a dictionary, each key a column name.

Two design choices are worth naming:

- `column` and `value` are **parameters**, not hard-coded. One function now filters any
  column instead of needing `filter_by_day`, `filter_by_item`, and so on.
- It builds and returns a **new list**. The original `sales` is untouched, so the raw data
  stays available for the next question.

````python
def filter_rows(rows: list[dict], column: str, value: object) -> list[dict]:
    """Return a new list of rows whose column equals value.

    The input table is not modified.
    """
    return [row for row in rows if row[column] == value]


monday = filter_rows(sales, "day", "Mon")
print("Monday rows:", len(monday))
for row in monday:
    print("  ", row["item"], row["price"])

print()
print("coffees   :", len(filter_rows(sales, "item", "coffee")))
print("students  :", len(filter_rows(sales, "customer_type", "student")))
print("original untouched:", len(sales))

# Filters compose, because each returns a table of the same shape:
student_drinks = filter_rows(filter_rows(sales, "customer_type", "student"), "category", "drink")
print("student drinks:", [row["item"] for row in student_drinks])
````

## Example 12 &mdash; Grouping rows

**Difficulty:** medium &nbsp;·&nbsp; **Teaches:** dict of lists, `setdefault`

Grouping turns a flat table into a dictionary whose values are **lists of rows**.

`groups.setdefault(key, [])` returns the list for that key, creating an empty one the first
time. It is the list-valued counterpart of `.get(key, 0)` from Example 8:

```python
groups.setdefault(key, []).append(row)
```

One line that means *"find the bucket for this key, making it if needed, then add the row."*

````python
def group_by(rows: list[dict], column: str) -> dict[object, list[dict]]:
    """Return rows grouped into a dictionary keyed by one column's value."""
    groups: dict[object, list[dict]] = {}
    for row in rows:
        key = row[column]
        groups.setdefault(key, []).append(row)
    return groups


by_day = group_by(sales, "day")
print("keys:", list(by_day))

for day, rows in by_day.items():
    items = ", ".join(row["item"] for row in rows)
    print(f"  {day}: {len(rows)} sales ({items})")

print()
by_category = group_by(sales, "category")
for category, rows in by_category.items():
    print(f"  {category:<6} total ${sum(r['price'] for r in rows):.2f}")
````

## Example 13 &mdash; Composing small functions into a report

**Difficulty:** medium &nbsp;·&nbsp; **Teaches:** decomposition, one job per function

This is the payoff of Examples 7&ndash;12. Rather than one long function, three small ones,
each with a single job:

| Function | Its one job |
|---|---|
| `column_values` | pull one column out of the table |
| `summarize` | compute statistics from a list of numbers |
| `format_summary` | turn statistics into text |

Because `summarize` takes a plain list, it knows nothing about coffee carts and can be
tested on `[1, 2, 3]`. Because `format_summary` only formats, changing the layout cannot
break the arithmetic. **Separating computing from reporting** is the habit to take away.

````python
def column_values(rows: list[dict], column: str) -> list:
    """Return one column of the table as a flat list."""
    return [row[column] for row in rows]


def summarize(values: list[float]) -> dict[str, float] | None:
    """Return count/lowest/highest/mean, or None when there is nothing to summarize."""
    if not values:
        return None
    return {
        "count": len(values),
        "lowest": min(values),
        "highest": max(values),
        "mean": sum(values) / len(values),
    }


def format_summary(stats: dict[str, float] | None) -> str:
    """Return a one-line report for the statistics."""
    if stats is None:
        return "no data"
    return (f"n={stats['count']}  range={stats['lowest']:.2f}-{stats['highest']:.2f}  "
            f"mean={stats['mean']:.2f}")


print(format_summary(summarize(column_values(sales, "price"))))
print(format_summary(summarize([])))

# Each stage is independently testable, with no coffee cart in sight:
assert summarize([1, 2, 3])["mean"] == 2
assert summarize([]) is None
print("stage tests passed")
````

## Example 14 &mdash; `TypeError` versus `ValueError`

**Difficulty:** medium &nbsp;·&nbsp; **Teaches:** choosing the right exception

The two exceptions mean different things, and the distinction helps whoever has to debug:

- **`TypeError`** &mdash; the *kind of object* is wrong. A string where a number belongs.
  Usually a **bug in the calling code**.
- **`ValueError`** &mdash; the type is right but the *value* is not allowed. A zero divisor.
  Usually **bad data**.

The `isinstance(b, bool)` check exists because in Python `bool` is a subclass of `int`, so
`isinstance(True, int)` is `True`. Without that guard, `safe_divide(10, True)` would
quietly return `10.0`.

````python
def safe_divide(a: float, b: float) -> float:
    """Return a / b.

    Raises:
        TypeError: If either argument is not a number.
        ValueError: If b is zero.
    """
    for name, value in (("a", a), ("b", b)):
        if isinstance(value, bool) or not isinstance(value, (int, float)):
            raise TypeError(f"{name} must be a number, got {type(value).__name__}")
    if b == 0:
        raise ValueError("b must not be zero")
    return a / b


print(safe_divide(10, 4))

for args in [(10, 0), ("10", 4), (10, None), (10, True)]:
    try:
        safe_divide(*args)
    except (TypeError, ValueError) as error:
        print(f"safe_divide{args} -> {type(error).__name__}: {error}")

print()
print("why the bool guard:", isinstance(True, int), "| True + True =", True + True)
````

## Example 15 &mdash; Return `None` instead of raising

**Difficulty:** medium &nbsp;·&nbsp; **Teaches:** choosing a data-quality policy

Example 14 raised on bad input. Sometimes that is wrong: when cleaning a thousand rows, one
malformed value should not stop the job.

This version **returns `None` for unusable input** and lets the caller decide. The
`try`/`except` catches the conversion failure and turns it into a value the program can
carry around.

Neither policy is universally correct. What matters is that the choice is **explicit,
documented, and consistent** &mdash; and the `float | None` annotation makes it visible in
the signature.

````python
def parse_price(text: object) -> float | None:
    """Return text converted to a price, or None when it cannot be used.

    Returns None -- rather than raising -- so that one bad row does not stop
    a whole batch. The caller is expected to count and report the rejects.
    """
    try:
        value = float(str(text).strip().lstrip("$"))
    except (TypeError, ValueError):
        return None
    if value < 0:
        return None
    return round(value, 2)


raw = ["3.50", "$2.75", " 1.00 ", "", "free", None, "-4.00", "2"]

cleaned, rejected = [], []
for item in raw:
    price = parse_price(item)
    if price is None:
        rejected.append(item)
    else:
        cleaned.append(price)

print("cleaned :", cleaned)
print("rejected:", rejected)
print(f"kept {len(cleaned)} of {len(raw)} values")
````

## Example 16 &mdash; Returning a new list instead of mutating

**Difficulty:** medium &nbsp;·&nbsp; **Teaches:** unannounced side effects

Both functions below "normalize" a list. Only one leaves the caller's data intact.

`normalize_in_place` uses `values[:] = ...`, which overwrites the contents of the **caller's
own list**. The original measurements are gone &mdash; unrecoverable, and no longer available
for comparison or audit.

`normalize` builds and returns a **new list**. The caller keeps both versions.

**Default to returning something new.** Mutate only when you mean to, and say so in the
name &mdash; as the next example does.

````python
def normalize(values: list[float]) -> list[float]:
    """Return a NEW list scaled so the largest value becomes 1.0."""
    largest = max(values)
    return [value / largest for value in values]


def normalize_in_place(values: list[float]) -> None:
    """Scale values in place. The caller's list is overwritten."""
    largest = max(values)
    values[:] = [value / largest for value in values]


original = [2.0, 4.0, 8.0]
scaled = normalize(original)
print("returned:", scaled)
print("original:", original, "<- still intact")
print("same object?", scaled is original)

print()
original = [2.0, 4.0, 8.0]
normalize_in_place(original)
print("after in-place:", original, "<- the raw numbers are gone")
````

## Example 17 &mdash; A matched pair: pure and in-place

**Difficulty:** medium &nbsp;·&nbsp; **Teaches:** naming and return conventions for mutation

When both behaviors are genuinely useful, provide **two functions** and let the names do the
explaining. Python's own library sets the pattern:

| Returns a new object | Mutates in place |
|---|---|
| `sorted(x)` | `x.sort()` |
| `reversed(x)` | `x.reverse()` |
| `with_row(table, row)` | `add_row(table, row)` |

The convention: the in-place version returns **`None`**. That is a feature &mdash; it stops
you writing `table = add_row(table, row)` and silently getting `None`.

````python
def with_row(table: list[dict], row: dict) -> list[dict]:
    """Return a NEW table with row appended. The input table is unchanged."""
    return table + [row]


def add_row(table: list[dict], row: dict) -> None:
    """Append row to table, modifying it in place.

    Returns:
        None, following the convention of list.append and list.sort.
    """
    table.append(row)


new_sale = {"id": 109, "item": "latte", "category": "drink", "price": 4.25,
            "day": "Thu", "customer_type": "staff"}

bigger = with_row(sales, new_sale)
print("new table:", len(bigger), "| original:", len(sales), "<- unchanged")

copy_of_sales = list(sales)
result = add_row(copy_of_sales, new_sale)
print("after add_row:", len(copy_of_sales), "| it returned:", result)

# Why returning None is helpful -- the mistake fails loudly instead of silently:
copy_of_sales = list(sales)
copy_of_sales = add_row(copy_of_sales, new_sale)
print("assigning the result gives:", copy_of_sales)
````

## Example 18 &mdash; Returning a list of problems

**Difficulty:** medium &nbsp;·&nbsp; **Teaches:** collecting results instead of stopping at the first error

`raise` reports **one** problem and stops. For validating a form or a data row, you usually
want **all** the problems at once, so the user can fix them in one pass.

The pattern: accumulate messages in a list and return it. An **empty list means valid**,
which reads naturally because an empty list is falsy:

```python
if not problems:
    ...
```

The function itself never prints and never raises &mdash; it just reports. That makes it easy
to test and easy to reuse behind a web form, a CSV loader, or a notebook.

````python
def validate_row(row: dict, required: list[str]) -> list[str]:
    """Return a list of problems with row. An empty list means the row is valid."""
    problems = []

    for field in required:
        if field not in row:
            problems.append(f"missing field {field!r}")
        elif row[field] in ("", None):
            problems.append(f"empty value for {field!r}")

    if "price" in row and isinstance(row["price"], (int, float)):
        if row["price"] < 0:
            problems.append("price cannot be negative")

    return problems


REQUIRED = ["id", "item", "price"]

candidates = [
    {"id": 201, "item": "latte", "price": 4.25},     # fine
    {"id": 202, "item": "", "price": -1.00},         # two problems
    {"id": 203, "price": 2.00},                      # missing item
]

for row in candidates:
    problems = validate_row(row, REQUIRED)
    if not problems:
        print(f"row {row['id']}: OK")
    else:
        print(f"row {row['id']}: {len(problems)} problem(s)")
        for problem in problems:
            print("   -", problem)
````

## Example 19 &mdash; Passing a function as an argument

**Difficulty:** medium &nbsp;·&nbsp; **Teaches:** higher-order functions

A function is an **object**, so it can be passed to another function just like a number or
a list. A function that accepts a function is called a **higher-order function**.

Notice the call: `apply_to_all(prices, add_tax)` &mdash; **no parentheses** on `add_tax`.
`add_tax` is the function itself; `add_tax(3.50)` would be its result. That distinction is
the whole idea.

One `apply_to_all` now covers every per-value transformation, instead of a separate
`add_tax_to_all`, `double_all`, `round_all`.

````python
from typing import Callable

def apply_to_all(values: list[float], func: Callable[[float], float]) -> list[float]:
    """Return a new list with func applied to every value."""
    return [func(value) for value in values]


def add_tax(price: float) -> float:
    return round(price * 1.08, 2)


prices = [3.50, 2.75, 1.75]

print(apply_to_all(prices, add_tax))          # our function
print(apply_to_all(prices, round))            # a BUILT-IN works too
print(apply_to_all(prices, lambda p: p * 2))  # a lambda works too

# Proof that the name and the call are different things:
print("the object:", add_tax)
print("the result:", add_tax(3.50))
````

## Example 20 &mdash; A function as a sort key

**Difficulty:** medium &nbsp;·&nbsp; **Teaches:** `key=`, lambdas in practice

`sorted()` takes a `key` function, calls it on every element, and sorts by the results.
This is the most common place a beginner passes a function as an argument.

For a table, the key receives a **whole row** and returns the value to sort on.

Two techniques worth stealing:

- **Sort on two columns** by returning a **tuple**: `(category, price)` sorts by category,
  then by price within it.
- **Reverse one column only** by negating it: `-row["price"]`.

````python
def show(rows, title):
    print(title)
    for row in rows:
        print(f"   {row['category']:<7}{row['item']:<8}{row['price']:>6.2f}")
    print()


show(sorted(sales, key=lambda row: row["price"]), "by price, cheapest first:")
show(sorted(sales, key=lambda row: -row["price"])[:3], "three most expensive:")
show(sorted(sales, key=lambda row: (row["category"], -row["price"])),
     "by category, then price descending:")

print("cheapest item:", min(sales, key=lambda row: row["price"])["item"])
````

## Example 21 &mdash; A function that builds a function

**Difficulty:** medium &nbsp;·&nbsp; **Teaches:** closures, function factories

`make_range_checker` **returns a function**. Note `return check` with no parentheses: it
returns the function object itself, not a result.

The returned function still remembers `minimum` and `maximum` even though
`make_range_checker` has already finished. A function that remembers values from where it
was **created** is a **closure**, and you can inspect what it captured.

Each call to the factory creates a **separate** closure, which is why `valid_age` and
`valid_price` do not interfere.

````python
def make_range_checker(minimum: float, maximum: float) -> Callable[[float], bool]:
    """Return a function that reports whether a value falls within [minimum, maximum]."""

    def check(value: float) -> bool:
        return minimum <= value <= maximum

    check.__name__ = f"in_{minimum}_to_{maximum}"
    return check                 # the function OBJECT, not a call


valid_age = make_range_checker(0, 120)
valid_price = make_range_checker(0.50, 10.00)

print(valid_age(34), valid_age(-5), valid_age(200))
print(valid_price(3.50), valid_price(0.10))

# Each closure kept its OWN captured values. Pair each free variable with its
# cell to see them by name (co_freevars is alphabetical, not signature order).
for checker in (valid_age, valid_price):
    captured = dict(zip(checker.__code__.co_freevars,
                        (cell.cell_contents for cell in checker.__closure__)))
    print(f"{checker.__name__:<16} captured {captured}")

# And it is an ordinary function, so it works as an argument:
print([p for p in [0.10, 3.50, 99.0] if valid_price(p)])
````

## Example 22 &mdash; Fixing arguments with `functools.partial`

**Difficulty:** medium &nbsp;·&nbsp; **Teaches:** `partial`, configured functions

Example 21 built a closure by hand. When the function **already exists** and you only want
to **fix some of its arguments**, the standard library does it for you.

`partial(apply_rate, rate=0.04)` returns a new callable that needs only `price`. The
original `apply_rate` is untouched and still usable.

Use `partial` for "same function, one argument pinned"; write a closure by hand when you
need to compute something at build time or attach metadata.

````python
from functools import partial

def apply_rate(price: float, rate: float) -> float:
    """Return price with a rate applied. An ordinary two-argument function."""
    return round(price * (1 + rate), 2)


georgia = partial(apply_rate, rate=0.04)
atlanta = partial(apply_rate, rate=0.089)

print(georgia(3.50), atlanta(3.50))
print("captured:", georgia.keywords, "| wraps:", georgia.func.__name__)

# One-argument callables, so they drop straight into Example 19's helper:
prices = [row["price"] for row in sales[:4]]
print("base   :", prices)
print("georgia:", apply_to_all(prices, georgia))
print("atlanta:", apply_to_all(prices, atlanta))

print("the original still takes two arguments:", apply_rate(3.50, 0.10))
````

## Example 23 &mdash; A dispatch table

**Difficulty:** medium &nbsp;·&nbsp; **Teaches:** functions in a dictionary

Because functions are values, they can be **dictionary values**. A dictionary mapping names
to functions is called a **dispatch table**, and it replaces a long `if`/`elif` chain.

The advantage is not just brevity. Adding an operation means adding **one dictionary
entry** &mdash; no existing code is edited, so nothing existing can break. The error message
can also list the valid options by reading the keys.

````python
def _mean(values): return sum(values) / len(values)
def _spread(values): return max(values) - min(values)

STATISTICS: dict[str, Callable[[list[float]], float]] = {
    "mean": _mean,
    "min": min,               # built-ins are objects too
    "max": max,
    "sum": sum,
    "spread": _spread,
}


def calculate(statistic: str, values: list[float]) -> float:
    """Apply a named statistic to values.

    Raises:
        ValueError: If the statistic is not one of STATISTICS.
        ValueError: If values is empty.
    """
    if statistic not in STATISTICS:
        raise ValueError(f"unknown statistic {statistic!r}; "
                         f"expected one of {sorted(STATISTICS)}")
    if not values:
        raise ValueError("values must not be empty")
    return round(STATISTICS[statistic](values), 2)


prices = [row["price"] for row in sales]
for name in STATISTICS:
    print(f"  {name:<7}{calculate(name, prices)}")

try:
    calculate("median", prices)
except ValueError as error:
    print(f"ValueError: {error}")
````

## Example 24 &mdash; Composing two functions

**Difficulty:** medium &nbsp;·&nbsp; **Teaches:** taking functions in *and* returning one

`compose` takes two functions and returns a **third** that runs them in sequence. It is both
a higher-order function (Example 19) and a factory (Example 21).

Read `combined(value)` as `f(g(value))`: the **inner** function `g` runs first, then `f`
takes its result. That right-to-left order matches the mathematical notation and is the
usual source of confusion &mdash; the demonstration below makes the order visible.

````python
def compose(f: Callable, g: Callable) -> Callable:
    """Return a function that applies g first, then f."""

    def combined(value):
        return f(g(value))

    combined.__name__ = f"{f.__name__}_after_{g.__name__}"
    return combined


def add_ten(n): return n + 10
def double(n): return n * 2


double_then_add = compose(add_ten, double)     # double runs FIRST
add_then_double = compose(double, add_ten)     # add_ten runs first

print(double_then_add.__name__, "of 5 ->", double_then_add(5))   # (5*2)+10 = 20
print(add_then_double.__name__, "of 5 ->", add_then_double(5))   # (5+10)*2 = 30

# Order matters -- and the composed function is ordinary, so it is reusable:
print(apply_to_all([1, 2, 3], double_then_add))
````

## Example 25 &mdash; A function that wraps another function

**Difficulty:** medium &nbsp;·&nbsp; **Teaches:** closures with state, and what a decorator is

The capstone: `memoize` takes a function and returns a **new function with the same
signature** that remembers previous answers in a `cache` dictionary. On a repeat call it
returns the stored result instead of recomputing.

Two attributes are attached to the wrapper &mdash; `cache` and `real_calls` &mdash; purely
so the demonstration can *prove* the caching happened. Functions are objects, so attaching
them costs nothing.

This combines everything in Part D &mdash; a function taken as an argument, a closure holding
state (`cache`), and a function returned as the result.

It also shows what a **decorator** is. `@memoize` above a `def` is simply shorthand for
`slow_square = memoize(slow_square)`. Nothing new &mdash; just syntax for wrapping.

````python
import time

def memoize(func: Callable) -> Callable:
    """Return a version of func that caches its results."""
    cache = {}                      # lives in the closure, one per wrapped function

    def wrapper(*args):
        if args in cache:
            return cache[args]      # a HIT: skip the real call entirely
        wrapper.real_calls += 1     # a MISS: do the work and remember it
        result = func(*args)
        cache[args] = result
        return result

    wrapper.__name__ = f"memoized_{func.__name__}"
    wrapper.cache = cache           # functions are objects -- expose the cache
    wrapper.real_calls = 0          # ...and a counter, so we can prove it works
    return wrapper


def slow_square(n):
    time.sleep(0.2)                 # pretend this is expensive
    return n * n


fast_square = memoize(slow_square)

start = time.time()
print("fast_square(12) ->", fast_square(12))    # a miss: takes 0.2s
print("fast_square(12) ->", fast_square(12))    # a hit: instant
print("fast_square(5)  ->", fast_square(5))     # a miss
elapsed = time.time() - start

print()
print(f"three calls took {elapsed:.1f}s, not 0.6s")
print("real calls that ran the body:", fast_square.real_calls)
print("cache:", fast_square.cache)
````

````python
# The @ syntax does exactly the same thing.

@memoize
def fibonacci(n):
    """Return the nth Fibonacci number."""
    return n if n < 2 else fibonacci(n - 1) + fibonacci(n - 2)

# `fibonacci` is now the WRAPPER, so even the RECURSIVE calls hit the cache.
print("fibonacci(30) =", fibonacci(30))
print("real calls    :", fibonacci.real_calls, "(one per distinct n, 0 through 30)")
print("cache entries :", len(fibonacci.cache))

# Without memoizing, the same recursion repeats work exponentially:
def plain_fibonacci(n, counter=None):
    counter["n"] += 1
    return n if n < 2 else plain_fibonacci(n - 1, counter) + plain_fibonacci(n - 2, counter)

counter = {"n": 0}
plain_fibonacci(25, counter)
print(f"un-memoized fibonacci(25) needed {counter['n']:,} calls "
      f"(memoized needs 26)")

# Proof the decorator is just a function call:
def plain(n):
    return n * 3

plain = memoize(plain)              # identical to writing @memoize above it
print(plain(4), plain(4), "| real calls:", plain.real_calls)
````
